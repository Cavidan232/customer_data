```python
# Şəhərlər üzrə ümumi xərcləməni hesablayırıq
customer_data.groupby("city")["total_spend"].sum()
```

```python
# Cins üzrə müştəri sayını tapırıq
customer_data["gender"].value_counts()
```

```python
#cins uzre orta yas
customer_data.groupby("gender")["age"].mean()
```

```python
#orta sifaris deyeri 250 den cox olanlar
customer_data[customer_data["avg_order_value"]>250]
```

```python
#her musterinin orta hesabla ne qeder xerclediyi
customer_data["spentperOrder"]=customer_data["total_spend"]/customer_data["purchase_count"]
customer_data["spentperOrder"]
```

```python
#en cox pul xercleyen 5 musteri
customer_data.sort_values("spentperOrder", ascending=False).head(5)[
    ["first_name", "last_name", "city", "purchase_count", "total_spend", "spentperOrder"]
]
```

```python
#en cox alis eden 5 musteri
customer_data.sort_values("purchase_count",ascending=False).head(5)[
    ["first_name","last_name","city","purchase_count","total_spend"]
]
```

```python
#Her seher ucun ortalama xercler
customer_data.groupby("city")["total_spend"].mean()
```

```python
#Her cins ucun ortalama xercler
customer_data.groupby("gender")["total_spend"].mean()
```

```python
#her cins ucun orta yas, umumi xerc, musteri sayi
customer_data.groupby("gender").agg(
    total_spend=("total_spend", "sum"),
    avg_age=("age", "mean"),
    customer_count=("customer_id", "count")
)
```

```python
# Şəhərlər üzrə müştəri sayı və ümumi xərci hesablayıb, ümumi xərcə görə sıralamaq
customer_data.groupby("city").agg(
    customer_size=("customer_id","size"),
    total_spend=("total_spend", "sum")
    
).sort_values(by="total_spend",ascending=False)
```

```python
# Şəhər və cins üzrə orta xərci müqayisə etmək
customer_data.pivot_table(
    index="city",
    columns="gender",
    values="total_spend",
    aggfunc="mean"
)
```

```python
# Aktiv alış edən və yüksək ümumi xərcə sahib müştəriləri tapmaq
active_customers=customer_data[
(customer_data["purchase_count"]>=10) & (customer_data["total_spend"]>=2000)
]
active_customers.sort_values("total_spend",ascending=False).head(10)
```

```python
#her sutundaki bos deyerlerin sayi
customer_data.isnull().sum()
```

```python
# missing deyeri olan sutunlar ve onlarin adlari
missing_columns = customer_data.isnull().sum()
missing_columns[missing_columns > 0].index
```

```python
# Total spend sütunundakı boş dəyərləri orta qiymətlə doldurmaq
customer_data["total_spend"]=customer_data["total_spend"].fillna(
    customer_data["total_spend"].mean()
)
customer_data["total_spend"].isnull().sum()
```

```python
# Dataset-də təkrarlanan sətirlərin sayını yoxlamaq
customer_data.duplicated().sum()
```

```python
#dublikatlari silmek
customer_data = customer_data.drop_duplicates()
customer_data.shape
```
